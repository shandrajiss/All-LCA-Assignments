```python
# To create list, tuple & dictionary and update, delete in dictionary.

Student_list=["Shandra", "Renee", "Jen", "Lavi"]
print("Student List:",Student_list)

student_tuple=("Shandra","Renee","Jen","Lavi")
print("Student Tuple:",student_tuple)

student_dict={100:"Shandra", 101:"Renee", 102:"Jen", 103:"Lavi"}
print("Student Dictionary:",student_dict)
```

```python
# Add a new element in dictionary.

student_dict[104]= "Helen"
print("After adding:",student_dict)
```

```python
# Update a student's name.

student_dict[101]="Ren"
print("After updating:",student_dict)
```

```python
# Delete a student.

student_dict.pop(100,None)
print("After deleting:",student_dict)
```
