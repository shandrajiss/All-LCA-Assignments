```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Step 1: Create a small dataset
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "Sales": [100, 150, 120, 180, 200],
    "Profit": [20, 30, 25, 40, 50],
    "Customers": [50, 60, 55, 75, 90]
}

# Step 2: Save the dataset as a CSV file
df = pd.DataFrame(data)
df.to_csv("sales_data.csv", index=False)

# Step 3: Load the CSV dataset
df = pd.read_csv("sales_data.csv")

print("Dataset:")
print(df)

# Step 4: Line plot
plt.figure(figsize=(7, 4))
plt.plot(df["Month"], df["Sales"], marker="o")
plt.title("Monthly Sales - Line Plot")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.tight_layout()
plt.show()

# Step 5: Bar plot
plt.figure(figsize=(7, 4))
plt.bar(df["Month"], df["Sales"], color="skyblue")
plt.title("Monthly Sales - Bar Plot")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.tight_layout()
plt.show()

# Step 6: Scatter plot
plt.figure(figsize=(7, 4))
plt.scatter(df["Sales"], df["Profit"], color="red")
plt.title("Sales vs Profit - Scatter Plot")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.grid(True)
plt.tight_layout()
plt.show()

# Step 7: Heatmap using Seaborn
plt.figure(figsize=(6, 4))
correlation = df[["Sales", "Profit", "Customers"]].corr()
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```
