```python
# Matrix Addition using Numpy.

import numpy as np

def sub(x,y):
    return x-y

def add(x,y):
    return x+y

def mul(x,y):
    return x*y

def div(x,y):
    return(x/y)

x=np.array ([[1,2,3],[4,5,6]])
y=np.array ([[7,8,9],[10,11,12]])

z=input (
    "Please select one of the operators:\n"
    "addition:+\n"
    "subtraction:-\n"
    "multiplication:*\n"
    "division:/\n"
)
if z=="+":
   print(add(x,y))

elif z=="-":
    print(sub(x,y))

elif z=="*":
    print(mul(x,y))

elif z=="/":
    print(div(x,y))

else:
    print("Invalid operator")
```
